## Scoping: what did your Romi data look like for forward speed?

```python
from control.matlab import step,tf
from numpy import *
from matplotlib.pyplot import *

tau = 0.05
SS = 0.9
noise = 0.1

model = control.matlab.tf([SS],[tau, 1])
ymodel,t = control.matlab.step(model)
for k in range(0,len(ymodel)):
    ymodel[k]+=(random.randint(-2,2))*noise
    
figure()
plot(t,ymodel,'k')
xlabel('Time (s)')
ylabel('Romi Speed (m/s)')
```
